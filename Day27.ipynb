```python
#np.where
where_result=np.where(numbers==5)
# get index position where the number 5 is

print(where_result)# gives index position of 5
print("number where",numbers[where_result])
```

```python
#row
original=np.array([[1,2],
      [3,4]
])

new_row=np.array([[5,6]])

added_row=np.vstack((original,new_row))#vertical stack
added_row
```

```python
#column add
original=np.array([[1,2],
      [3,4]
])

new_col=np.array([[5],[6]])
added_col=np.hstack((original,new_col))#horizontal stack
added_col
```
