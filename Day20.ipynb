```python
class BankAccount:
  def __init__(self, name, pin, balance):
    self.name = name
    self._type = "Saving"
    self.__pin = pin
    self.__balance = balance

  def get_balance(self, pin):
    if pin == self.__pin:
      return self.__balance
    else:
      print("Incorrect PIN")

  def _update_balance(self, amount):
    self.__balance += amount
```

```python
class SavingAccount(BankAccount):
 def add_interest(self):
  balance= self.get_balance(self.pin)

  if isinstance(balance, (int, float)):
    interest=balance * 0.05
    self._update_balance(balance*0.05)
    print(f"Interest added {interest}")
  else:
    print("Not Updated.Balance:", balance)
```

```python
#Write a program to create an ATM class with a private PIN and methods to verify and change the PIN securely.

class ATM:
  def __init__(self, account_name, pin):
    self.name = account_name
    self.__pin = pin
  def verify_pin(self, entered_pin):
    if entered_pin == self.__pin:
      print("PIN verified successfully!")
      return True
    else:
      print("Access Denied: Incorrect PIN!")
      return False
  def change_pin(self, current_pin, new_pin):
    if not self.verify_pin(current_pin):
      print("Failed to change PIN.")
      return
    if len(str(new_pin)) == 4 and isinstance(new_pin, int):
      self.__pin = new_pin
      print("PIN changed successfully!")
    else:
      print("Invaid PIN Format")

```

```python
#Write a program where Student and Teacher classes override the role() method inherited from the Person class.


from IPython.core.magics import namespace
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def role(self):
    print(f"{self.name} (Age: {self.age}) is a General Person.")

class Student(Person):
  def __init__(self, name, age, student_id):
    super().__init__(name, age)
    self.student_id = student_id
  def role(self):
    print(f"{self.name} (ID: {self.student_id}) is a Student.")

class Teacher(Person):
  def __init__(self, name, age, subject):
    super().__init__(name, age)
    self.subject = subject
  def role(self):
    print(f"{self.name} is a Teacher teaching '{self.subject}'")


person = Person("Robin", 45)
student = Student("Penny", 24, "ST4044")
teacher = Teacher("Clint", 33, "Geology")
person.role()
student.role()
teacher.role()
```

```python
#Write a program to create an abstract Payment class and implement payment processing for different payment methods.


from abc import ABC, abstractmethod

class Payment(ABC):
  def __init__(self, amount):
    self.amount = amount
  @abstractmethod
  def process_payment(self):
    pass

class CreditCard(Payment):
  def __init__(self, amount, card_number):
    super().__init__(amount)
    self.card_number = card_number
  def process_payment(self):
    if self.amount > 0:
      print(f"Payment of Rs.{self.amount} successful.")
      return True
    else:
      print("Payment failed: Invalid amount.")
      return False

class Esewa(Payment):
  def __init__(self, amount, esewa_id):
    super().__init__(amount)
    self.esewa_id = esewa_id
  def process_payment(self):
    if self.amount > 0:
      print(f"Payment of Rs.{self.amount} via {self.esewa_id} successful.")
      return True
    else:
      print("Payment failed: Invalid amount.")
      return False

class Khalti(Payment):
  def __init__(self, amount, khalti_id):
    super().__init__(amount)
    self.khalti_id = khalti_id
  def process_payment(self):
    if self.amount > 0:
      print(f"Payment of Rs.{self.amount} via {self.khalti_id} successful.")
      return True
    else:
      print("Payment failed: Invalid amount.")
      return False

```

```python
#Write a program where SavingsAccount and CurrentAccount classes override the interest_rate() method of the Account class.

class Account:
  def __init__(self, account_holder, balance):
    self.account_holder = account_holder
    self.balance = balance
  def interest_rate(self):
    rate = 0.0
    print(f"Standard Account Interest Rate: {rate}")
    return rate

class SavingsAccount(Account):
  def __init__(self, account_holder, balance):
    super(). __init__(account_holder, balance)

  def interest_rate(self):
    rate = 6.5
    print(f"Savings Account Interest Rate for {self.account_holder}: {rate}")

class CurrentAccount(Account):
  def __init__(self, account_holder, balance):
    super().__init__(account_holder, balance)

  def interest_rate(self):
    rate = 1.0
    print(f"Current Account Interest Rate for {self.account_holder}: {rate}")
```

```python
acc = Account("User", 7000)
savings = SavingsAccount("Ali", 25000)
current = CurrentAccount("Business", 75000)

acc.interest_rate()
savings.interest_rate()
current.interest_rate()

```

```python
#Write a program to create an abstract Animal class and print different sounds using child classes.

from abc import ABC, abstractmethod

class Animal(ABC):
  @abstractmethod
  def make_sound(self):
    pass

class Dog(Animal):
  def make_sound(self):
    print("Woof! Woof!!")

class Cat(Animal):
  def make_sound(self):
    print("Meow! Meow!!")

class Duck(Animal):
  def make_sound(self):
    print("Quack! Quack!!")

```

```python

dog = Dog()
dog.make_sound()


cat = Cat()
cat.make_sound()


duck = Duck()
duck.make_sound()
```
