```python
#Operator Overloading
class Point:
  def __init__(self,x,y):
    self.x=x
    self.y=y

  def __add__(self,other):
     x=self.x + other.x
     y=self.y + other.y
     return Point(x,y)

  def __sub__(self,other):
     x=self.x - other.x
     y=self.y -  other.y
     return Point(x,y)

  def __mul__(self,other):
     x=self.x * other.x
     y=self.y * other.y
     return Point(x,y)



  def __str__(self):
    return f"({self.x},{self.y})"




```

```python
p1 = Point(1, 2)
p2 = Point(3,4)
```

```python
p3=p1+p2
print(p3)
```

```python
p4= p1 - p2
print(p4)
```

```python
#division
class Points:
  def __init__(self,x, y):
    self.x = x
    self.y = y

  def __truediv__(self, number):
    return Points(self.x/number, self.y/number)
  def __str__(self):
    return f"{self.x}, {self.y}"

```

```python
p1 = Points(15,21)
p2 = p1 / 2
print(p2)


```

```python
#reminder
class product:
  def __init__(self,stock):
    self.stock = stock

  def __mod__(self, copy):
    return product(self.stock % copy)

  def __str__(self):
    return f"{self.stock}"

prod = product(1045)
remaining = prod % 10
print(f"Remaining from the stock = {remaining}")
```

```python
#equal to
class Data:
  def __init__(self, id, balance):
    self.id = id
    self.balance = balance
  def __eq__(self, other):
    return self.id == other.id
  def __str__(self):
    return f"{self.id} = {self.balance}"

a1 = Data("100", 300 )
a2 = Data("100", 300)
print(a1 == a2)

```

```python
#getitems
class List:
  def __init__(self,data):
    self.data = data

  def __getitem__(self, index):
    return self.data[index]

numbers = List([10, 20, 30])

print(numbers[0])
print(numbers[2])
```
