```python
def factorial (n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)
```

```python
factorial(7)
```

```python
#Fibonacci numbers
def fibonacci(n):
  if n<=1:
   return 1
  else:
   return (fibonacci(n-1)+fibonacci(n-2))
```

```python
def myWrapper (func):
  def myInnerFunc():
    print("Inner wrapper.")
    func()
  return myInnerFunc
```

```python
@myWrapper#decorator
def myFunc():
  print("Hello, world!")
```
