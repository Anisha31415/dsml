```python
class Parent:
  def myMethod(self):# self instance method
   print("Calling parent method")
```

```python
class Employee:
  def __init__(self, nm, sal):
    self.name=nm
    self.salary=sal

  def getName(self):
    return self.name

  def getSalary(self):
    return self.salary
```

```python
class SalesOfficer(Employee):
  def __init__(self, nm, sal, inc):
    super().__init__(nm, sal)
    self.incentive = inc
  def getSalary(self):
    return self.salary + self.incentive
```

```python
e1=Employee("Anisha", 90000)
print("Total salary of {} is Rs.{}". format(e1.getName(), e1.getSalary()))
```

```python
s1 = SalesOfficer("Anu", 19000, 20000)
print("Total salary of {} is Rs.{}". format(s1.getName(), s1.getSalary()))
```

```python
class Example:
  def __init__(self):
    self.public_var=1 #public variable
    self.__private_var=2 # private variable
    self._protected_var=3 #protected variable

  def public_method(self):
    print("Public method")
    print("Private variable:", self.__private_var)

  def __private_method(self):
    print("Private method")

  def _protected_method (self):
    print("Protected method")

```

```python
obj= Example()
```

```python
#Accessing public members
print(obj.public_var)
obj.public_method()
```

```python
print(obj._protected_var)
obj._protected_method
```

```python
#Accessing public members
print(obj._Example__private_var)
obj._Example__private_method()
```

Create a class BankAccount that:

1.Allows a user to create an account by providing a name and an initial balance. Let balance be private variable.

2.Include methods to deposit, withdraw and check balance. The withdrawal should not allow an account to go negative

3.Uses Inheritance to create a subclass SavingsAccount that adds an interest rate. Implement a method to calculate interest on the balance and update it accordingly.

```python
class BankAccount:
  def __init__(self,name, initial_balance):
    self.name=name
    self.__balance=initial_balance

  def deposit(self, amount):
    if amount > 0:
      self.__balance += amount
      print(f"Successfully deposited: Rs.{amount}.\nNew Balance: Rs.{self.__balance}")
    else:
      print("Deposit amount must be greater than zero.")

  def withdraw(self, amount):
    if amount <= 0:
      print("Withdrawal amount must be greater than zero.")
    elif amount > self.__balance:
      print(f"Transaction failed! Insufficient funds. Current balance: Rs.{self.__balance}")
    else:
      self.__balance -= amount
      print(f"Successfully withdrawn: Rs.{amount}.\nNew Balance: Rs.{self.__balance}")

  def check_balance(self):
    print(f"Account Holder: {self.name} | Balance: Rs.{self.__balance}")

  def get_balance(self):
    return self.__balance
```

```python
class SavingsAccount(BankAccount):
  def __init__(self, name, balance, interest_rate):
    super().__init__(name, balance)
    self._account_type = "Saving"
    self.__rate = interest_rate

  def add_interest(self):
    current_balance = self.get_balance()
    interest = current_balance * self.__rate
    new_balance = current_balance + interest
    self._BankAccount__balance = (new_balance)
    print(f"Applied Interest: Rs.{interest}. New Balance: Rs.{self.get_balance()}")

  def account_info(self):
    print("Account Holder: ", self.name)
    print("Account Type: ", self._account_type)
```

```python
acc= SavingsAccount("Ami", 500, 0.05)
acc.account_info()
```

```python
from abc import ABC, abstractmethod
class Animal(ABC):
  @abstractmethod
  def make_sound(self):
    pass

class Dog(Animal):
  def make_sound(self):
    print("Woof!")

class Cat(Animal):
  def make_sound(self):
    print("Meow!")


dog = Dog()
dog.make_sound()

cat = Cat()
cat.make_sound()

```
