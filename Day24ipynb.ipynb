```python
# Recursion: Recursion is a programming technique where a function calls itself, until a stopping condition

# Recursion
#Recursion is a programming technique where a function calls itself, until a stopping condition
#Base case (stopping condition)
#Recursive case

# Factorial number

# 5! = 5*4*3*2*1
# fact(n) * fact(n-1)


def condition(n):
  if n==0 or n==1: #base case
    return 1
  return n*condition(n-1) #recursive case
```

```python
condition(5)
```

```python
class A:
  def show(self):
    print("A")

class B(A):
  def show(self):
     print("B")
     super().show()

class C(A):
  def show(self):
    print("C")
    super().show()

class D(B,C):
  def show(self):
    print("D")
    super().show()
```

```python
obj=D()
obj.show()
```

```python
#iterator and generator
#every generator is an iterator but not every iterator is a generator.Both are tools are used to implemment lazy evaluation
```

```python
class Count:
  def __init__(self, limit):
    self.num=1
    self.limit=limit

  def __iter__(self):
    return self

  def __next__(self):
    if self.num<=self.limit:
      val=self.num
      self.num+=1
      return val
    else:
      raise StopIteration
```

```python
counter=Count(5)
```

```python
for i in counter:
  print(i)
```

```python
#generator is a simpler way to create an iterator using the yield keyword.
def count(limit):

   num=1

   while num <= limit:
     yield num
     num += 1
```

```python
counter= count(5)
```

```python
for i in counter:
  print(i)
```

```python
def numbers(): #Only one value is generated at a time, making generators memory-efficient for large datasets.
  for i in range(1,6):
    yield i
g=numbers()
print(next(g))
print(next(g))
```
