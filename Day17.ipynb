**Class Task**

Create a Python class named Calculator that demonstrates the use of different types of attributes and methods.

Requirements:

1.Create a class attribute named company and assign it any company name.

2.Use the__init__() method to:
Ask the user to enter two numbers.
Ask the user to enter an operator (+,-,*,/).

3.Create an instance method named calculate() that performs the appropriate operation based on the operator entered by the user.

4.Create a class method named show_company() that displays the company name.

5.Create a static method named welcome() that displays a welcome message.

6.Create an object of the class and call all the required methods.


```python
class Calculator:
  #Class Attribute
  company = "XYZ"

  def __init__(self):
    #INSTANCE ATTRIBUTES
    self.num1 = float(input("Enter first number:"))
    self.num2 = float(input("Enter second number"))
    self.operator = input("Enter operator (+, -, *, /):")

  #instance method
  def calculate(self):

    if self.operator=="+":
       print("Result", self.num1 + self.num2)

    elif self.operator=="-":
       print("Result", self.num1 - self.num2)

    elif self.operator=="*":
       print("Result", self.num1 * self.num2)

    elif self.operator=="/":
       print("Result", self.num1 / self.num2)

    else:
       print("Invalid operator")

 #class method
  @classmethod
  def show_company(cls):
    print(f"The Company name is {cls.company}")

 #static method
  @staticmethod
  def welcome():
    print("Welcome")


```

```python
ca= Calculator()
ca.calculate()

```

```python
ca.show_company()
```

```python
ca.welcome()
```

Create a Python class named Student.

Requirements:

1.Create a class attribute named school.

2.Use the init() method to take:

Student name

Student marks

3.Create an instance method named
check_grade() that displays:

A Grade: Marks ≥ 80

B Grade: Marks ≥ 60 and < 80

C Grade: Marks ≥ 40 and < 60

Fail: Marks < 40

4.Create a class method to display the school name.

5.Create a static method to display a welcome message.

6.Create an object and display the student's grade


```python
class Student:
  #Class Attribute
  school= "XYZ"

  def __init__(self):
    #INSTANCE ATTRIBUTES
    self.name = input("Enter the name:")
    self.marks = float(input("Enter the marks"))

  #instance method
  def check_grade(self):

    if self.marks>=80:
       print("Grade A")

    elif self.marks>=60 and self.marks<80:
       print("Grade B")

    elif self.marks>=40 and self.marks<60:
       print("Grade C")

    elif self.marks<40:
       print("Fail")

    else:
       print("Invalid ")

 #class method
  @classmethod
  def show_school(cls):
    print(f"The School name is {cls.school}")

 #static method
  @staticmethod
  def welcome():
    print("Hello")


```

```python
Student.show_school()
```

```python
Student.welcome()

```

```python
#create object
std = Student()

#instance method
std.check_grade()
```

**Create  A Python class named Bank Account.**

Requirements:

1.Create a class attribute named bank_name.

2.Use the init() method to take:

Account holder name

Initial balance

3.Create the following instance methods:
deposit() → Ask the user for an amount and add it to the balance.

withdraw() → Ask the user for an amount and deduct it from the balance if sufficient balance is available
.
show_balance() → Display the current balance.

4.Create a class method to display the bank name.

5.Create a static method to display banking rules or a welcome message.

6.Create an object and test all methods.


```python
class BankAccount:
  #Class Attribute
  bank = "NIC"

  def __init__(self):
    #INSTANCE ATTRIBUTES
    self.name = str(input("Enter the name:"))
    self.balance = float(input("Enter the initial balance"))

  #instance method
  def deposit(self):
     depo=float(input("Enter the amount to deposit"))
     self.balance += depo
     print(f"Successfully deposited: {depo}")
     print(f"New balance: Rs.{self.balance}")

  def withdraw(self):
    withd=float(input("Enter the amount to be withdrawn"))
    self.balance -= withd
    print(f"Successfully withdrawn: {withd}")
    print (f"New Balance:Rs.{self.balance}")

  def show_balance(self):
   print(f"Account Holder Name:{self.name} | Current balance :Rs.{self.balance}")


  @classmethod
  def show_bank(cls):
    print(f"The bank name is {cls.bank}")

 #static method
  @staticmethod
  def welcome():
    print(f"Welcome to {BankAccount.bank}")


```

```python
#class Method
BankAccount.show_bank()

```

```python
#static method
BankAccount.welcome()
```

```python
#create object
ba = BankAccount()
ba.deposit()
```

```python
ba.withdraw()
```

```python
ba.show_balance()
```
