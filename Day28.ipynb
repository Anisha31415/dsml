```python
#Element_wise Matrix operations
import numpy as np
matrix_1=np.array([
    [1,2],
    [3,4]
     ])
matrix_1
```

```python
matrix_2=np.array([
    [4,6],
    [7,8]

])
matrix_2
```

```python
result=matrix_1 + matrix_2 #element wise addition
result
```

```python
matrix_sub=matrix_1- matrix_2 #element wise subtraction
matrix_sub
```

```python
#matrix multiplication size
result_mux=matrix_1* matrix_2 #wrong concept
result_mux
```

```python
result_mux = matrix_1 @ matrix_2  #right way to multiply
result_mux
```

```python
result_mux=np.matmul(matrix_1, matrix_2) #alternative way
result_mux
```

```python
#matrix division not important
result_div=matrix_1/matrix_2
result_div
```

```python
#elements wise operations with scalars
print(matrix_1)
print("/////////")
scalar=2
result_add=matrix_1+scalar
print(result_add)
```

```python
print(matrix_1)
print("/////////")
scalar=2
result_mux=matrix_1 *scalar
print(result_mux)
```

```python
#broadcasting  -allows  2D array to add with 1D array.
matrix_3 = np.array([
    [1,2],
    [3,4]
])
array_1 = np.array([5,6])#added to rows
result = matrix_3 + array_1
result
```

```python
#Determinants - use for matrix conversion and inverse
```

```python
#computing the determinant
#we can calculate  the determinant of a matrix using linalg.det() function
A=np.array([
    [1,2],
    [3,4]
])
det=np.linalg.det(A)
print(det)
```

```python
#inverse
#We can calulate the inverse of a matrix using the linalg.inv()function
inv_A=np.linalg.inv(A)
print(inv_A)
```

```python
#A* A-1=I
```

```python
print(A@inv_A)
```

```python
X = np.array([
    [1,1],
    [2,1],
    [3,1]
])

y = np.array([2,3,5])

beta=np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
print("slope(m)",beta[0])
```

```python
print("Intercept(c):", beta[1])
```
