```python
def area (*args):
 if len(args)==2:
  l,w=args
  print("2D Area =", l*w)
 elif len(args)==3:
  l,w,h =args
  print("3D volume =", l*w*h)
 else:
   print("invalid number of argument")



```

```python
area(5,4)
area(4,5,6)
```

```python
area(2)
```
