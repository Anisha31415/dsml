```python
#class work
#create 5*5 matrix
a=np.array([
    [1,2,3,4,5],
    [5,6,7,5,4],
    [3,4,5,6,7],
    [5,2,1,4,6],
    [3,2,6,7,8]
  ])
a.ndim
```

```python
print(a.shape)
print(a)
```

```python
# Calculate matrix rank also called no.of dimensions
rank = np.linalg.matrix_rank(a)
print("Matrix Rank:", rank)  # Output: 2 (Row 2 is a multiple of Row 1)
```
