```python
import numpy as np

A=np.array((
     [1,2],
     [3,4]
))
eigenvalues,  eigenvectors=np.linalg.eig(A)
print(eigenvalues)
print(eigenvectors)
```

```python
#eigenvector drawbacks= works on square not rectangle
#SVD-Singular Value Decomposition- decompose(rotate,stretch,rotate)
#A=UEVt=
#E summation e and transpose of V
U, S, V = np.linalg.svd(A)
print("U matrix:\n", U)
print("sigma values:", S)
print("V matrix:\n", V)
```

```python
print(U @ U.T)

```
