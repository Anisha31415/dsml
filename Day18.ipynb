```python
#Hybrid Inheritance
class Base:
  def super(self):
    pass

class ChildA(Base):
  def first_child(self):
    pass

class ChildB(Base):
  def second_child(self):
    pass

class GrandChild(ChildA, ChildB):
  def grand(self):
    pass
```

```python
a = GrandChild()
a.super()
a.first_child()
a.second_child()
a.grand()
```

```python
#Class Task
#Implement methods in each class and show a TeachingAssistant object can access features from both Student and Teacher.

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def show_person(self):
    print(f"Name: {self.name}  Age: {self.age}")

class Student(Person):
  def __init__(self, name, age, student_id):
    Person.__init__(self, name, age)
    self.student_id = student_id
  def study(self):
    print("Student ID: ", self.student_id)

class Teacher(Person):
  def __init__(self, name, age, subject):
    Person.__init__(self, name, age)
    self.subject = subject
  def teach(self):
    print("Subject: ", self.subject)

class TeachingAssistant(Student, Teacher):
  def __init__(self, name, age, student_id, subject, department, salary):
    Student.__init__(self, name, age, student_id)
    Teacher.__init__(self, name, age, subject)
    self.department = department
    self.salary = salary
  def assist(self):
    print("Department: ", self.department)
    print("Salary: ", self.salary)

```

```python
p1 = TeachingAssistant("Mark", 25, "A1076", "Maths", "DSML", 20000)
p1.show_person()
p1.study()
p1.teach()
p1.assist()

```

```python
class OTTPSubscription:
  def __init__(self, subscription_id, plan, total_payment):
    self.id = subscription_id
    self.plan = plan
    self.total_payment = total_payment
  def subscription(self):
    print(f"Subscriber with {self.id} is subscribed to the {self.plan}")
  def unsubscribe(self):
    print(f"Subscriber with {self.id} unsubscribed to the {self.plan}")
```

```python
class PremiumSubscription(OTTPSubscription):
  def __init__(self, subscription_id, plan, total_payment, screens):
    super().__init__(subscription_id, plan, total_payment)
    self.max_screen = screens
  def set_max_screen(self):
    print(f"Maximum screen set to {self.max_screen} in the premium plan.")



```

```python
obj=PremiumSubscription(2000, "Pro", 13000, 9)
obj.set_max_screen()
```

```python
obj.subscription()
obj.unsubscribe()

```

```python
#Create classes Teacher and Researcher, each with their own methods. Create a class Professor that inherits from both classes and demonstrates the functionality of both parents.

class Teacher:
  def __init__(self, subject):
    self.subject = subject
  def teach(self):
    print(f"Teaching {self.subject}")

class Researcher:
  def __init__(self, field):
    self.field = field
  def research(self):
    print(f"Conducting research on {self.field}")

class Professor(Teacher, Researcher):
  def __init__(self, name, subject, field):
    Teacher.__init__(self, subject)
    Researcher.__init__(self, field)
    self.name = name
  def display(self):
    print("Professor Name: ", self.name)
    self.teach()
    self.research()


prof1 = Professor("Dr. Smith", "Microbiology", "Hantavirus")
prof1.display()

```

```python
#Create a class Shape with a method display(). Create two child classes Circle and Rectangle that inherit from Shape and implement their own area calculation methods.

from re import S
import math

class Shape:
  def display(self):
    print("This is a geometric shape.")

class Circle(Shape):
  def __init__(self, radius):
    self.radius = radius
  def area(self):
    a = math.pi * (self.radius**2)
    print(f"Area of circle: {a}")

class Rectangle(Shape):
  def __init__(self, length, breadth):
    self.length = length
    self.breadth = breadth
  def area(self):
    a = self.length * self.breadth
    print(f"Area of rectangle: {a}")
```

```python
circle1 = Circle(5)
circle1.display()
circle1.area()
```

```python
rect1 = Rectangle(6,7)
rect1.display()
rect1.area()
```
