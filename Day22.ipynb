```python
#magic method
class BankCustomer:
  def __init__(self,name,age):
    self.name=name
    self.age=age

 #called by repr()
  def __repr__(self):
    return f"{self.name} is {self.age} years old"

#called by str()
  def __str__(self):
    return f"{self.name} is {self.age} years old"

# called by len()
  def __len__(self):
    return self.age
```

```python
customer=BankCustomer("John" , 20)
```

```python
#__str__
print(customer)
```

```python
#repr
print(repr(customer))
```

```python
#len
print(len(customer))
```
